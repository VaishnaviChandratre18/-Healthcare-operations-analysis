```python
#Admissions trend over time (by month)
df['Admission Month'] = df['Date of Admission'].dt.strftime('%b')

month_order = ['Jan','Feb','Mar','Apr','May','Jun',
               'Jul','Aug','Sep','Oct','Nov','Dec']

df['Admission Month'] = pd.Categorical(df['Admission Month'], categories=month_order, ordered=True)

result = df.groupby('Admission Month')['Billing Amount'].mean().sort_values(ascending=False)
print(result)
```

```python
#Age Distributuion
import pandas as pd
import matplotlib.pyplot as plt
plt.hist(df['Age'],color='mediumseagreen',bins=15,edgecolor='white')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')     #Patient age is fairly evenly spread across the dataset with no strong concentration in any particular age group.
plt.show()
```

```python
condition_counts = df['Medical Condition'].value_counts()

plt.bar(condition_counts.index, condition_counts.values, color='darkgreen')
plt.ylim(9000, 9350) 
plt.xlabel('Medical Condition')
plt.ylabel('Number of Patients')
plt.title('Patient Count by Medical Condition')
plt.xticks(rotation=45)
plt.show()

# Observation:All six medical conditions (Arthritis, Diabetes, Hypertension, Obesity, 
# Cancer, Asthma) have consistent patient counts.
```

```python
insurance_billing = df.groupby('Insurance Provider')['Billing Amount'].mean().sort_values(ascending=False)

plt.barh(insurance_billing.index, insurance_billing.values, color='lightgreen')
plt.xlim(25000, 25700) 
plt.ylabel('Insurance Provider')
plt.xlabel('Average Billing Amount')
plt.title('Average Billing Amount by Insurance Provider')
plt.show()

# Observation: Average billing amount of Medicare leads while United healthcare is least invested in terms of insurance provider
```

```python
plt.boxplot(df['Length of Stay'])
plt.ylabel('Length of Stay (days)')
plt.title('Length of Stay Distribution')
plt.show()
#There are no Outliers present in the data
```

*Key Insights

-The patient age distribution is spread across multiple age groups, with a higher concentration in middle-aged patients.
-Billing amounts vary significantly, indicating differences in treatment costs and healthcare services.
-Insurance providers show noticeable differences in average billing amounts.
-Test result categories are relatively balanced across the dataset.
-A very weak relationships is between Age,Billing Amount, and Length of Stay, suggesting these variables are largely independent in a linear sense.
-No Outliers are found in the data.
-Arthrities and diabities are amongst top conditions patients are getting admiitted for while ashthama and cancer among the least.

*Recommendations
-Investigate the February decline in billing/admissions to identify possible causes such as seasonal demand changes, reduced patient visits, staffing constraints, or operational disruptions.
-Introduce monthly performance monitoring and alert mechanisms to detect sudden drops in revenue or patient activity early.
-Plan targeted patient engagement and marketing initiatives during historically weaker months to stabilize revenue trends.
-Review hospital resource allocation and scheduling during low-performing periods to improve operational efficiency and maintain service quality.
